#### Zadanie 1: Klasyfikacja tekstowa

Masz kolumnę z opisami produktów.

Użyj apply, aby stworzyć kolumnę 'kategoria_dlugosc' z wartościami:

'krótki' (<50 znaków),
'średni' (50-100),
'długi' (>100).
Następnie zlicz słowa w opisie.

```python
import pandas as pd

df = pd.DataFrame({
    'id': range(1, 6),
    'opis': [
        'Małe krzesło',
        'Eleganckie biurko z drewna dębowego z szufladami',
        'Kompaktowy laptop do pracy biurowej z procesorem i7, 16GB RAM i dyskiem SSD 512GB, idealny do codziennych zadań',
        'Lampa',
        'Wygodna kanapa trzysegmentowa w kolorze szarym z funkcją spania'
    ]
})
```

```python
def classify_sentence(text):
   n = len(text)
   if n < 50:
      return 'krótki'
   elif n < 100:
      return 'średni'
   return 'długi'

# df['opis'].apply(lambda x: print(f"wyświetlam wiersz: {x}") or x)
# _ = df['opis'].apply(lambda x: print(f"wyświetlam wiersz: {x}") or x)
df['kategoria_dlugosc'] = df['opis'].apply(lambda x: classify_sentence(x))

# ---------------
# Liczba słów
# -----------------------
df['liczbe_slow'] = df['opis'].apply(lambda x: len(x.split()))

df
```
